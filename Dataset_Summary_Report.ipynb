# Dataset Summary Report — Student Dataset

This notebook explores `Day7_Student_Dataset.csv` using Pandas: loading the
data, checking its basic structure, and pulling out descriptive statistics
and a few observations about the data quality.

```python
import pandas as pd

# In Colab: upload Day7_Student_Dataset.csv via the Files panel first
df = pd.read_csv("Day7_Student_Dataset.csv")
print("Dataset loaded.")
```

## 1. First and Last Rows

`head()` shows the first few rows, `tail()` shows the last few — a quick way
to see what the data looks like without printing the whole thing.

```python
df.head()
```

```python
df.tail()
```

## 2. Random Sample

`sample()` pulls random rows instead of rows from the top/bottom, which is
useful for a less biased look at the data.

```python
df.sample(5, random_state=1)
```

## 3. Shape, Columns, and Index

- `shape` gives (rows, columns)
- `columns` lists the column names
- `index` shows how the rows are labeled

```python
print("Shape (rows, columns):", df.shape)
print("\nColumns:", list(df.columns))
print("\nIndex:", df.index)
```

## 4. Data Types

`dtypes` shows the data type Pandas assigned to each column.

```python
df.dtypes
```

## 5. info()

`info()` gives a combined summary: column names, non-null counts, and dtypes
in one view. This is also the fastest way to spot missing values.

```python
df.info()
```

## 6. describe()

`describe()` generates summary statistics (count, mean, std, min, quartiles,
max) for the numeric columns.

```python
df.describe()
```

## 7. Checking for Missing Values

`isnull().sum()` counts how many missing (NaN) values each column has.

```python
df.isnull().sum()
```

## 8. Observations

- The dataset has **25 rows** and **7 columns**.
- Columns: `Student_ID`, `Name`, `Age`, `Course`, `Marks`,
  `Attendance_Percentage`, `City`.
- Data types: `Student_ID` and `Age` are integers, `Marks` and
  `Attendance_Percentage` are floats, and `Name`, `Course`, `City` are text
  (object) columns.
- Missing values: `Marks`, `Attendance_Percentage`, and `City` each have
  **1 missing value**. The other columns have no missing data.
- `Course` has 3 distinct values (Data Science, AI, Computer Science) based
  on a quick look at the rows above.
- `Marks` and `Attendance_Percentage` both range roughly from the high 60s
  to the mid-to-high 90s, based on the `describe()` output.
